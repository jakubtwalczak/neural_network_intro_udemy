# Wektoryzacja tekstu.

Jest to proces zakodowania danych tekstowych za pomocą liczb, co pozwala na wprowadzenie tekstu do sieci neuronowej (sieć nie działa bowiem na danych tekstowych).

## Podział tekstu na słowa.

NLP operuje określeniem "token" na pojedynczy znak lub wyraz będący częścią sekwencji tekstowej.

```python
from tensorflow.keras.preprocessing.text import text_to_word_sequence
# ww. pozwala podzielić i wystandaryzować tekst - pominąć znaki interpukcyjne i zamienić wszystkie litery na małe

text = 'Keras is a high-level neural networks API, written in Python and capable of running on top of TensorFlow, CNTK, or Theano.'

tokens = text_to_word_sequence(text)
tokens
```

## Kodowanie one-hot.

Nie jest tożsame z kodowaniem zero-jedynkowym zmiennych kategorycznych. Polega ono na przedstawieniu każdego słowa w tekście jako liczby całkowitej.

Funkcja one_hot() z Keras wykorzystuje dwa parametry:
- input_text - tekst do zakodowania,
- n - liczba tokenów w tekście (rozmiar słownika).

Tekst jest przez tę funkcję kodowany do listy indeksów słów o rozmiarze n. Jest to opakowanie funkcji hashującej. Rozmiar słownika określa przestrzeń hashująca, z której słowa są hashowane - aby zapobiec kolizjom, najlepiej byłoby, gdyby przestrzeń ta była większa niż słownik.

Poniżej przykłady hashowania słów:

```python
hash('network')
```

```python
hash('network') % 100 # operacja typu modulo może stanowić indeks słownika
```

```python
hash('network') # za każdym razem dla takiego samego tekstu jest zwracana ta sama wartość
```

```python
hash('neural')
```

```python
# aby takich operacji nie dokonywać ręcznie, mamy wysokopoziomową funkcję one_hot()

from tensorflow.keras.preprocessing.text import one_hot

words = set(tokens)
one_hot_tokens = one_hot(text, round(len(words) * 1.3))
one_hot_tokens
```

Dla każdego słowa przyporządkowany jest dzięki temu indeks całkowito-liczbowy.

## Kodowanie hashing trick.

Funkcja hashing_trick() różni się tym, że pozwala dodatkowo wskazać jako parametr funkcję hashującą (domyślnie None).

```python
from tensorflow.keras.preprocessing.text import hashing_trick

hashing_trick(text, round(len(words) * 1.3), hash_function='md5')
```

## Tokenizer.

Klasa Tokenizer() pozwala zamienić każdy tekst na sekwencję liczb całkowitych w taki sposób, że każda liczba całkowita jest indeksem tokenu w słowniku. Tokenem zazwyczaj jest pojedyncze słowo. Tokenizer() indeksuje tokeny od liczby 1; 0 jest zarezerwowanym indeksem i nie może zostać przypisane do żadnego słowa.

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer() # tworzymy oczywiście instancję klasy

# popracujemy z listą zdań odwzorowujących uprzejme komentarze w social mediach
samples = ['Great picture!', 'Nice view', 'Good to see you :)', 'Good picture!', 'Good']

tokenizer.fit_on_texts(samples) # dopasowujemy klasę na ww. próbkach

tokenizer.index_word # atrybut zwraca słowa jako wartości słownika z indeksami typu integer
```

```python
tokenizer.word_counts # częstość występowania słowa w próbkach
```

```python
tokenizer.document_count # zlicza wszystkie dokumenty - sekwencje tekstu w liście
```

Po dopasowaniu Tokenizera do danych treningowych można go użyć do kodowania dokumentów - zarówno ze zbioru treningowego, jak i testowego.

Używamy funkcji texts_to_matrix(), aby stworzyć wektor dla każdego dokumentu, o długości równej liczbie unikalnych słów we wszystkich dokumentach.

```python
print(samples)
print(tokenizer.index_word)
print(tokenizer.texts_to_matrix(samples))
```
